### Méthode 1 : df1 — Supprimer les lignes avec des valeurs manquantes

```python
df1 = df.dropna()
print("Taille originale :", df.shape)
print("Taille après suppression des lignes avec NA :", df1.shape)
print("Lignes perdues :", df.shape[0] - df1.shape[0])
```

### Méthode 2 : df2 — Supprimer la colonne la plus manquante puis les lignes restantes

```python
# Identifier la colonne avec le plus de valeurs manquantes
most_missing_col = df.isnull().sum().drop('Date').idxmax()
print(f"Colonne avec le plus de NA : {most_missing_col} ({df[most_missing_col].isnull().sum()} valeurs manquantes)")

df2 = df.drop(most_missing_col, axis=1).dropna()
print("\nTaille après suppression de '{}' puis des lignes NA : {}".format(most_missing_col, df2.shape))
print("Vérification NA restantes :", df2.isnull().sum().sum())
```

### Méthode 3 : df3 — Remplacer les valeurs manquantes par la médiane

```python
# Créer df3 avec GOLD et WTI CRUDE
df3 = df[['GOLD', 'WTI CRUDE']].copy()

print("Valeurs manquantes avant :")
print(df3.isnull().sum())

# Remplacer les NA par la médiane
for col in df3.columns:
    median_val = df3[col].median()
    df3[col].fillna(median_val, inplace=True)
    print(f"\nMédiane de {col} : {median_val}")

print("\nValeurs manquantes après :")
print(df3.isnull().sum())
df3.head()
```

## 4. Normalisation Min-Max

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

# Appliquer MinMaxScaler sur df3
df3_scaled = pd.DataFrame(
    scaler.fit_transform(df3),
    columns=df3.columns
)

print("Données normalisées (Min-Max) :")
df3_scaled.describe()
```

```python
# Visualiser avant/après normalisation
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].set_title('Avant normalisation')
df3.plot(kind='hist', bins=50, alpha=0.7, ax=axes[0])
axes[0].set_xlabel('Prix')

axes[1].set_title('Après normalisation Min-Max')
df3_scaled.plot(kind='hist', bins=50, alpha=0.7, ax=axes[1])
axes[1].set_xlabel('Valeur normalisée')

plt.tight_layout()
plt.show()
```

## 5. Normalisation Standard (StandardScaler)

```python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()

df3_std = pd.DataFrame(
    std_scaler.fit_transform(df3),
    columns=df3.columns
)

print("Données normalisées (StandardScaler) :")
df3_std.describe()
```

## 6. Encodage — One-Hot Encoding (création d'une variable catégorielle)

> Le dataset commodity_futures ne contient que des variables numériques. 
> Pour illustrer le One-Hot Encoding, nous créons une variable catégorielle à partir des données.

```python
# Créer une variable catégorielle : tendance du GOLD (Hausse / Baisse / Stable)
df_encoded = df[['Date', 'GOLD']].dropna().copy()
df_encoded['GOLD_Change'] = df_encoded['GOLD'].pct_change()

def categorize_trend(change):
    if pd.isna(change):
        return 'Stable'
    elif change > 0.01:
        return 'Hausse'
    elif change < -0.01:
        return 'Baisse'
    else:
        return 'Stable'

df_encoded['Trend'] = df_encoded['GOLD_Change'].apply(categorize_trend)
print("Répartition de la variable Trend :")
print(df_encoded['Trend'].value_counts())
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False)
one_hot = encoder.fit_transform(df_encoded[['Trend']])

encoded_cols = pd.DataFrame(one_hot, columns=encoder.get_feature_names_out(['Trend']))
print("Colonnes après One-Hot Encoding :")
print(encoded_cols.head(10))
```

```python
# Fusionner avec le DataFrame original
df_final_encoded = pd.concat([df_encoded.reset_index(drop=True), encoded_cols], axis=1)
print("\nDataset après One-Hot Encoding :")
df_final_encoded.head(10)
```

# TP3 : Régression Linéaire

> Nous allons prédire le prix du **GOLD** en utilisant les autres commodités comme variables explicatives.

## 1. Préparation des données

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, median_absolute_error, r2_score
```

```python
# Charger et nettoyer les données
df = pd.read_csv("commodity_futures.csv")
df = df.drop(columns=['Date'])

# Supprimer les lignes avec des valeurs manquantes
df = df.dropna()

print("Dimensions après nettoyage :", df.shape)
df.head()
```

---
## Régression Linéaire Simple
## $$f(X) = aX + b$$

> On essaie d'estimer le prix du GOLD en utilisant une seule variable : **SILVER** (la plus corrélée).

---

### 7. Définir X et Y

```python
X = df[['SILVER']]
y = df[['GOLD']]

print("X shape :", X.shape)
print("y shape :", y.shape)
```

### 8. Diviser les données en train/test (70%-30%)

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=4)

print("X_train :", X_train.shape)
print("X_test :", X_test.shape)
print("y_train :", y_train.shape)
print("y_test :", y_test.shape)
```

### 9. Entraîner le modèle et afficher les coefficients

```python
model = LinearRegression()
model.fit(X_train, y_train)

print("Coefficient (a) :", model.coef_)
print("Intercept (b) :", model.intercept_)
```

### 10. Prédire et évaluer les performances

$$ R^2 = 1 - \frac{\sum (y_{vrai} - y_{pred})^2}{\sum (y_{vrai} - \overline{y_{vrai}})^2} $$
$$ MAE = \frac{1}{m} \sum |y_{vrai} - y_{pred}| $$
$$ MSE = \frac{1}{m} \sum (y_{vrai} - y_{pred})^2 $$
$$ RMSE = \sqrt{MSE} $$

```python
y_pred = model.predict(X_test)

print('R2 = ', r2_score(y_test, y_pred))
print('MSE = ', mean_squared_error(y_test, y_pred))
print('RMSE = ', np.sqrt(mean_squared_error(y_test, y_pred)))
print('MAE = ', mean_absolute_error(y_test, y_pred))
print('MedianAE = ', median_absolute_error(y_test, y_pred))
```

### 11. Visualiser la droite de régression

```python
plt.figure(figsize=(10, 6))
plt.scatter(X_test, y_test, color='blue', alpha=0.3, label='Données réelles')
plt.plot(X_test.sort_values('SILVER'), model.predict(X_test.sort_values('SILVER')), 
         color='red', linewidth=2, label='Régression linéaire')
plt.xlabel("SILVER")
plt.ylabel("GOLD")
plt.title("Régression Linéaire Simple : GOLD ~ SILVER")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

---
## Régression Linéaire Multiple
## $$f(X_1, X_2, ...) = a_0 + a_1 X_1 + a_2 X_2 + ... + a_n X_n$$

> On utilise toutes les commodités pour prédire le GOLD.

---

### 12. Séparer features et target, diviser en train/test

```python
X = df.drop('GOLD', axis=1)
y = df[['GOLD']]

print("X shape :", X.shape)
print("y shape :", y.shape)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=4)
```

### 13. Entraîner le modèle multiple

```python
model2 = LinearRegression(fit_intercept=True)
model2.fit(X_train, y_train)

print("Score R² (train) :", model2.score(X_train, y_train))
print("\nIntercept :", model2.intercept_)
print("\nCoefficients :")
for name, coef in zip(X.columns, model2.coef_[0]):
    print(f"  {name:25s} : {coef:.4f}")
```

### 14. Évaluer les performances

```python
y_pred2 = model2.predict(X_test)

print('R2 = ', r2_score(y_test, y_pred2))
print('MSE = ', mean_squared_error(y_test, y_pred2))
print('RMSE = ', np.sqrt(mean_squared_error(y_test, y_pred2)))
print('MAE = ', mean_absolute_error(y_test, y_pred2))
print('MedianAE = ', median_absolute_error(y_test, y_pred2))
```

### 15. Visualisation : Prédictions vs Valeurs réelles

```python
plt.figure(figsize=(10, 6))
plt.scatter(y_test, y_pred2, alpha=0.3, color='green')
plt.plot([y_test.min().values[0], y_test.max().values[0]], 
         [y_test.min().values[0], y_test.max().values[0]], 
         color='red', linewidth=2, linestyle='--', label='Ligne parfaite')
plt.xlabel("Valeurs réelles (GOLD)")
plt.ylabel("Valeurs prédites (GOLD)")
plt.title("Régression Linéaire Multiple : Prédictions vs Réel")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

---
## Normalisation des Données + Régression
---

### 16. Normaliser avec StandardScaler et entraîner le modèle

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model3 = LinearRegression()
model3.fit(X_train_scaled, y_train)

print("Score R² (train, normalisé) :", model3.score(X_train_scaled, y_train))
```

### 17. Évaluer les performances après normalisation

```python
y_pred3 = model3.predict(X_test_scaled)

print('R2 = ', r2_score(y_test, y_pred3))
print('MSE = ', mean_squared_error(y_test, y_pred3))
print('RMSE = ', np.sqrt(mean_squared_error(y_test, y_pred3)))
print('MAE = ', mean_absolute_error(y_test, y_pred3))
print('MedianAE = ', median_absolute_error(y_test, y_pred3))
```

### 18. Comparaison des modèles

```python
results = pd.DataFrame({
    'Modèle': ['Régression Simple (SILVER)', 'Régression Multiple', 'Régression Multiple (Normalisée)'],
    'R2': [
        r2_score(y_test, model.predict(X_test.values.reshape(-1,1) if X_test.shape[1]==1 else X_test)),
        r2_score(y_test, y_pred2),
        r2_score(y_test, y_pred3)
    ],
    'RMSE': [
        np.sqrt(mean_squared_error(y_test, model.predict(X_test))),
        np.sqrt(mean_squared_error(y_test, y_pred2)),
        np.sqrt(mean_squared_error(y_test, y_pred3))
    ],
    'MAE': [
        mean_absolute_error(y_test, model.predict(X_test)),
        mean_absolute_error(y_test, y_pred2),
        mean_absolute_error(y_test, y_pred3)
    ]
})

results
```
